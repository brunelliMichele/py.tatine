```python
import matplotlib.pyplot as plt

# Dati
models = [
    "EfficientNetB0",
    "EfficientNetB4",
    "EfficientNet_merged",
    "ResNet 50",
    "ResNet finetuned\n(L2 + CE, 5 ep)",
    "Dino",
    "CLIP",
    "CLIP_fine_tuned",
    "VGG16"
]

scores = [
    17.319587628865982,
    23.02405498281787,
    23.02405498281787,
    30.58419243986254,
    51.134020618556704,
    37.11340206185567,
    734.639175257732,
    0.6872852233676976,
    3.848797250859107
]

# Sort models by performance
sorted_data = sorted(zip(models, scores), key=lambda x: x[1], reverse=True)
models_sorted = [x[0] for x in sorted_data]
scores_sorted = [x[1] for x in sorted_data]

# Plot
plt.figure(figsize=(10, 6))
bars = plt.bar(models_sorted, scores_sorted, color='skyblue', edgecolor='black')

# Add values on top of bars
for bar, score in zip(bars, scores_sorted):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.015,
             f'{score:.3f}', ha='center', va='bottom', fontweight='bold')

# Aesthetics
plt.title('Model Performance During Competition', fontsize=14, fontweight='bold')
plt.ylabel('Precision@K (log scale)')
plt.yscale('log')  # scala logaritmica
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```
