```python
import matplotlib.pyplot as plt

# Data
models = ['EfficientNet B0', 'EfficientNet B4', 'EfficientNet Merged', 'ResNet-50',
          'DINO', 'CLIP', 'VGG16']
precision_scores = [0.72, 0.73, 0.57, 0.65, 0.88, 0.835, 0.085]

# Sort models by performance
sorted_data = sorted(zip(models, precision_scores), key=lambda x: x[1], reverse=True)
models_sorted = [x[0] for x in sorted_data]
scores_sorted = [x[1] for x in sorted_data]

# Plot
plt.figure(figsize=(10, 6))
bars = plt.bar(models_sorted, scores_sorted, color='skyblue', edgecolor='black')

# Add values on top of bars
for bar, score in zip(bars, scores_sorted):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.015,
             f'{score:.3f}', ha='center', va='bottom', fontweight='bold')

# Aesthetics
plt.title('Model Performance Comparison (Precision@K)', fontsize=14, fontweight='bold')
plt.ylabel('Precision@K')
plt.ylim(0, 1.0)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

```
